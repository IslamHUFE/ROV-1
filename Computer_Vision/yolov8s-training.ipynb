```python
import os

# Configuration
DATASET_PATH = "Datasets"
MODEL_PATH = "yolov8n.pt"

# Disable AMP for stability
os.environ['YOLO_AMP'] = '0'
def simple_fine_tune():
    """
    Simple fine-tuning with dataset in working directory
    """
    print("🚀 Starting Simple Fine-Tuning")
    print("=" * 40)
    
    # 1. Copy dataset to working directory
    dataset_path = DATASET_PATH
    
    # 2. Load model
    print("🔧 Loading model...")
    model = YOLO(str(MODEL_PATH))
    
    # 4. Simple training configuration
    print("🎯 Starting training...")
    
    model.train(
        data='finaldata/data.yaml',
        epochs=50,
        batch=4,
        imgsz=640,
        lr0=0.001,
        patience=10,
        device='cpu',
        val=True,  
        workers=0,
        save=True,
        name='yolov8n_finetuned',
        exist_ok=True,
        verbose=True
    )
    
    return model

# Main execution
if __name__ == "__main__":
    print("🎯 YOLOv8s Simple Fine-Tuning")
    print("==============================")
    
    try:
        model = simple_fine_tune()
        
        # Save the final model
        model_path ='fine_tuned_model_4_2_2026.pt'
        model.save(model_path)
        print(f"✅ Training completed!")
        print(f"💾 Model saved: {model_path}")
        
    except Exception as e:
        print(f"❌ Training failed: {e}")
```

```python
import subprocess
import os
import time

def avi_to_mp4_hd(avi_path, mp4_path=None, resolution=None, bitrate='high'):
    """
    Convert AVI to high-quality HD MP4
    
    Args:
        avi_path: Path to input AVI file
        mp4_path: Path to output MP4 file
        resolution: Target resolution (e.g., '1920x1080', '1280x720', '3840x2160')
        bitrate: Quality preset ('low', 'medium', 'high', 'ultra')
    """
    if mp4_path is None:
        filename = os.path.basename(avi_path)
        filename_no_ext = os.path.splitext(filename)[0]
        mp4_path = f"videos/{filename_no_ext}_hd.mp4"
    
    # Bitrate presets (higher = better quality, larger file)
    bitrate_presets = {
        'low': {
            'video_bitrate': '2M',
            'audio_bitrate': '128k',
            'crf': '28'
        },
        'medium': {
            'video_bitrate': '5M',
            'audio_bitrate': '192k',
            'crf': '23'
        },
        'high': {
            'video_bitrate': '10M',
            'audio_bitrate': '256k',
            'crf': '18'
        },
        'ultra': {
            'video_bitrate': '20M',
            'audio_bitrate': '320k',
            'crf': '14'
        },
        'lossless': {
            'video_bitrate': '0',
            'audio_bitrate': '320k',
            'crf': '0'
        }
    }
    
    preset = bitrate_presets.get(bitrate, bitrate_presets['high'])
    
    # FFmpeg command for HD quality
    command = [
        'ffmpeg',
        '-i', avi_path,
        '-c:v', 'libx264',          # H.264 codec
        '-preset', 'slow',          # Better compression, slower encoding
        '-crf', preset['crf'],      # Quality (0=lossless, 51=worst)
        '-b:v', preset['video_bitrate'],  # Video bitrate
        '-maxrate', f"{int(preset['video_bitrate'][:-1])*2}M",  # Max bitrate
        '-bufsize', f"{int(preset['video_bitrate'][:-1])*4}M",  # Buffer size
        '-c:a', 'aac',              # Audio codec
        '-b:a', preset['audio_bitrate'],  # Audio bitrate
        '-movflags', '+faststart',  # Optimize for web streaming
        '-pix_fmt', 'yuv420p',      # Pixel format for compatibility
        '-profile:v', 'high',       # H.264 profile
        '-level', '4.1',            # H.264 level
        '-tune', 'film',            # Optimize for film content
        '-vsync', '1',              # Frame rate sync
        '-threads', '0',            # Use all CPU threads
        '-y'                        # Overwrite output
    ]
    
    # Add resolution scaling if specified
    if resolution:
        command.extend(['-vf', f'scale={resolution}:flags=lanczos'])
    
    command.append(mp4_path)
    
    print(f"🎬 Converting to HD MP4: {avi_path}")
    print(f"   Quality: {bitrate}")
    print(f"   Resolution: {resolution or 'Original'}")
    print(f"   Output: {mp4_path}")
    
    start_time = time.time()
    
    try:
        # Run FFmpeg with progress output
        process = subprocess.Popen(
            command,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            universal_newlines=True,
            encoding='utf-8',
            errors='replace'
        )
        
        # Display progress
        for line in process.stdout:
            if 'frame=' in line or 'time=' in line:
                print(f"   {line.strip()}", end='\r')
        
        process.wait()
        
        if process.returncode == 0 and os.path.exists(mp4_path):
            elapsed_time = time.time() - start_time
            file_size = os.path.getsize(mp4_path) / (1024*1024*1024)  # GB
            print(f"\n✅ HD Conversion complete in {elapsed_time:.1f}s")
            print(f"   File size: {file_size:.2f} GB")
            print(f"   Bitrate: {preset['video_bitrate']}")
            print(f"   CRF: {preset['crf']}")
            return mp4_path
        else:
            print(f"❌ Conversion failed!")
            return None
            
    except FileNotFoundError:
        print("❌ FFmpeg not found. Please install FFmpeg:")
        print("   Windows: Download from https://ffmpeg.org/")
        print("   Ubuntu: sudo apt install ffmpeg")
        print("   Mac: brew install ffmpeg")
        return None
    except Exception as e:
        print(f"❌ Error: {e}")
        return None

# Usage examples
avi_to_mp4_hd(
    avi_path='detection_results/video_detection_mp4/istockphoto-1059514008-640_adpp_is.avi',
    mp4_path='detection_results/video_detection_mp4/underwater_hd2.mp4',
    resolution='1920x1080',  # 1080p HD
    bitrate='high'          # High quality
)

```

```python
import cv2
import numpy as np

def test_and_visualize(model, image_path, confidence=0.3):
    """Test model and visualize results with bounding boxes"""
    
    # Run prediction
    results = model.predict(image_path, conf=confidence)
    
    # Load original image
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # Create plot
    plt.figure(figsize=(12, 8))
    plt.imshow(image)
    
    # Plot bounding boxes
    for box in results[0].boxes:
        # Get box coordinates
        x1, y1, x2, y2 = box.xyxy[0].cpu().numpy()
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        class_name = model.names[class_id]
        
        # Draw rectangle
        plt.gca().add_patch(plt.Rectangle((x1, y1), x2-x1, y2-y1, 
                                        fill=False, edgecolor='red', linewidth=2))
        # Add label
        plt.text(x1, y1-10, f'{class_name} {confidence:.2f}', 
                bbox=dict(boxstyle="round,pad=0.3", facecolor='red', alpha=0.7),
                fontsize=8, color='white')
    
    plt.axis('off')
    plt.title(f'Hieroglyph Detection Results (confidence ≥ {confidence})')
    plt.tight_layout()
    plt.show()
    
    return len(results[0].boxes)

# Test the function
test_image = 'Datasets/under water debris detection.v1i.yolov8/test/images/GettyImages-1227891345-1000x667_jpg.rf.f2da3b84c768edb8d0f6e29d966cc9c7.jpg'  # Replace with actual path
detections = test_and_visualize(fine_tuned_model, test_image)
print(f"🔍 Detected {detections} objects in the test image")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
import os

def load_mistral():
    model_name = "mistralai/Mistral-7B-Instruct-v0.2"
    
    print("Loading Mistral model...")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    
    # Add padding token if missing
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        device_map="auto",
        torch_dtype=torch.float16,
        low_cpu_mem_usage=True
    )
    print("Mistral loaded successfully!")
    return model, tokenizer

def save_mistral_locally():
    try:
        model, tokenizer = load_mistral()
        local_path = "./mistral_model"
        
        # Create directory if it doesn't exist
        os.makedirs(local_path, exist_ok=True)
        
        # Save model and tokenizer
        model.save_pretrained(local_path)
        tokenizer.save_pretrained(local_path)
        print(f"Mistral saved locally to {local_path}")
        
    except Exception as e:
        print(f"Error saving model: {e}")
        return False
    return True

def load_local_model():
    local_path = "./mistral_model"
    
    # Check if local model exists
    if not os.path.exists(local_path):
        print("Local model not found. Downloading first...")
        if not save_mistral_locally():
            raise Exception("Failed to download model")
    
    print("Loading local Mistral model...")
    tokenizer = AutoTokenizer.from_pretrained(local_path)
    model = AutoModelForCausalLM.from_pretrained(local_path)
    
    # Ensure padding token
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
        
    print("Local model loaded successfully!")
    return model, tokenizer

# Run setup
if __name__ == "__main__":
    print("Setting up Mistral model...")
    save_mistral_locally()
```

```python
import json

def lookup_gardiner_values(input_string, json_file_path):
    """
    Look up Gardiner code values from a JSON file
    Handles comma-separated input with repeated codes
    """
    # Load the JSON file
    with open(json_file_path, 'r') as f:
        gardiner_dict = json.load(f)
    
    # Split by commas and/or spaces, then clean up
    import re
    codes = re.split(r'[,\s]+', input_string.strip())
    codes = [code.strip() for code in codes if code.strip()]
    
    # Look up each code in the dictionary (preserving duplicates)
    values = []
    for code in codes:
        if code in gardiner_dict:
            values.append(gardiner_dict[code])
        else:
            values.append(f"[{code}]")  # Mark unknown codes
    
    # Return as a comma-separated string
    return ", ".join(values)

# Example with your exact input
input_codes = ", ".join(unique_classes)
json_path = "hieroglyphics.json"

input = lookup_gardiner_values(input_codes, json_path)
print(f"Input: {input_codes}")
print(f"Output: {input}")
```

```python
def gardiner_translation(incomplete_sentence, max_length=2000):
    model, tokenizer = load_local_model()
    
    # FIX: Set padding token if it doesn't exist
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token
    
    # System prompt setting up the archaeologist persona
    system_prompt = """You are deciphering ancient Egyptian hieroglyphics carved into temple walls. 
Your mission: Use given words only to create coherent English sentences that reveal 
the secrets of ancient Egypt. The words:"""

    
    # Combine system prompt with the incomplete sentence
    full_prompt = system_prompt + incomplete_sentence
    
    # Add instruction for continuation
    full_prompt += "Complete the translation:"
    
    # Tokenize with padding token now set
    inputs = tokenizer(
        full_prompt,
        return_tensors='pt',
        padding=True,
        truncation=True,
        max_length=512
    )
    
    output = model.generate(
            input_ids=inputs['input_ids'],
            max_new_tokens=100,
            temperature=0.7,
            do_sample=True,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )
    # Decode and return
    completed_text = tokenizer.decode(output[0], skip_special_tokens=True)
    
    # Extract only the generated part (after the prompt)
    generated_part = completed_text.replace(full_prompt, "").strip()
    return generated_part

# Example usage
incomplete = input
result = gardiner_translation(incomplete)
print(f"Available characters: {incomplete}")
print(f"Archaeologist's translation: {result}")
```
